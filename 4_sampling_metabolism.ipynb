## Sampling standard conditions

```python
# Sampling all cells

results_file_name = 'results_new/results_std_corrected_12h_PEAKS.dat'

std = pd.read_csv(results_file_name, sep = '\t')

folder = 'results_new/genome_scale_samples_no_exchange_PEAKS/'
i = 0
for cell in std[i:].iterrows():
    c,n,g, sd = cell[1][2], cell[1][3], cell[1][4], cell[1][5]
    model = cobra.io.load_json_model('models/iJN678_std_aminoacids.json')
    model.reactions.EX_hco3_e.bounds = (-1000, -c)
    model.reactions.RBPC.bounds = (c, 1000)

    model.reactions.EX_nh4_e.bounds = (-1000,-n)
    model.reactions.EX_co2_e.lower_bound = 0
    model.reactions.BIOMASS_Ec_SynAuto.bounds = (g-sd, g+sd)
    model.reactions.BIOMASS_Ec_SynHetero.bounds = (0,0)
    model.reactions.BIOMASS_Ec_SynMixo.bounds = (0,0)

    model.reactions.EX_o2_e.lower_bound = -1000
    model.reactions.EX_so4_e.lower_bound = -200
    model.reactions.EX_photon_e.lower_bound = -23
    model.reactions.Diff_nh4_e.bounds = (0,1000)
    model.solver = 'glpk'
    try: 
        a = cobra.sampling.sample(model, n = 1000 , processes=4)
        a.to_csv(folder + 'std_1000samples_cell'+str(i)+'.dat', index=None, sep='\t')
    except:
        print('not feasible')
        pass

    i += 1
    
```

```python
light_corr = 1
model = cobra.io.load_json_model('models/iJN678.json')
model.reactions.EX_o2_e.lower_bound = 0
model.reactions.EX_glc__D_e.lower_bound = 0
model.reactions.EX_photon_e.lower_bound = -9 * light_corr
model.reactions.EX_co2_e.lower_bound = -1.26* light_corr
model.reactions.EX_hco3_e.lower_bound = 0
model.reactions.EX_no3_e.lower_bound = 0
model.reactions.EX_so4_e.lower_bound = -0.569 * light_corr
model.reactions.EX_nh4_e.lower_bound = -33* light_corr
ATP_m = cobra.Reaction(id = 'ATPM', name = 'ATP maintenance', 
                       lower_bound=0.13, upper_bound=0.13)
ATP_m.add_metabolites({model.metabolites.atp_c:-1,model.metabolites.h2o_c:-1,
                    model.metabolites.adp_c:1, model.metabolites.h_c:1,
                    model.metabolites.pi_c:1})
model.add_reactions([ATP_m])
model.objective = model.reactions.BIOMASS_Ec_SynAuto
model.optimize()
```
